```python
def load_detailed_data(file_path):
    df = pd.read_excel(file_path)
    df = df.drop(df.columns[13:22], axis=1)
    # Normalize column names
    df.columns = (
        df.columns.astype(str)
        .str.strip()
        .str.lower()
        .str.replace(" ", "_")
    )

    # Convert string "nan" → real NaN
    df = df.replace(["nan", "NaN", ""], pd.NA)

    return df
```

```python
df = load_detailed_data("Data/Graphite/graphite_import.xlsx")
df.to_excel("Data/Graphite/graphite_import.xlsx", index=False)
```
